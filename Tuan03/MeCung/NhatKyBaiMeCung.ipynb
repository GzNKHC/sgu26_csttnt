# Nhật ký thực hiện bài tập
## Bài toán mê cung – DFS & BFS

**Sinh viên:** …………………………………

**Môn học:** Cơ sở Trí tuệ Nhân tạo

**Mục tiêu:** Ghi lại quá trình phân tích, cài đặt và kiểm thử thuật toán tìm đường trong mê cung.

## Ngày 1 – Phân tích bài toán
- Đề bài yêu cầu tìm đường đi từ vị trí S đến ô A trong mê cung.
- Mê cung được biểu diễn dưới dạng lưới 2D.
- Quyết định mô hình hóa mê cung thành đồ thị, mỗi ô là một đỉnh.
- Lựa chọn hai thuật toán tìm kiếm: DFS và BFS.

## Ngày 2 – Thiết kế dữ liệu vào
- Dữ liệu được đọc từ file văn bản.
- Dòng đầu chứa số hàng và số cột.
- Các dòng tiếp theo biểu diễn mê cung với ký hiệu:
  - S: điểm bắt đầu
  - A: điểm đích
  - 0: ô trống
  - 1: tường

## Ngày 3 – Cài đặt thuật toán BFS
- Sử dụng hàng đợi (queue).
- Duyệt theo chiều rộng.
- Lưu lại nút cha để truy vết đường đi.
- BFS đảm bảo tìm được đường đi ngắn nhất.

## Ngày 4 – Cài đặt thuật toán DFS
- Sử dụng ngăn xếp (stack).
- Duyệt theo chiều sâu.
- Có thể không cho kết quả tối ưu nhưng dễ cài đặt.
- So sánh kết quả với BFS.

## Ngày 5 – Kiểm thử và đánh giá
- Chạy chương trình với nhiều mê cung khác nhau.
- BFS cho kết quả ổn định và tối ưu.
- DFS có thể đi vòng nhưng vẫn tìm được lời giải.
- Chương trình cho phép chuyển đổi giữa BFS và DFS bằng cách comment/uncomment.

## Nhận xét chung
- Bài toán giúp hiểu rõ bản chất các thuật toán tìm kiếm.
- Việc đọc dữ liệu từ file giúp chương trình linh hoạt.
- Có thể mở rộng bằng thư viện networkx để trực quan hóa đồ thị.
