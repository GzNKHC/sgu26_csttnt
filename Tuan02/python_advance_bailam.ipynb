## 1. Thao tác cơ bản với thư mục

Phần này minh họa cách:
- Lấy thư mục làm việc hiện tại
- Lấy thư mục tạm của hệ điều hành
- Nối và tạo đường dẫn thư mục

```python
import os
import tempfile

cwd = os.getcwd()
print(cwd)

tmpdir = tempfile.gettempdir()
print(tmpdir)

mytmpdir = os.path.join(tmpdir, "foobar")
print(mytmpdir)
```

**Ghi chú:**
- `os.getcwd()` trả về thư mục hiện tại đang làm việc
- `tempfile.gettempdir()` trả về thư mục tạm của hệ điều hành
- `os.path.join()` giúp nối đường dẫn độc lập hệ điều hành (Windows/Linux)

## 2. Tạo và liệt kê thư mục

```python
os.makedirs(os.path.join(tmpdir, "foobar", "plop", "toto"), exist_ok=True)
os.listdir(mytmpdir)
```

**Ghi chú:**
- `os.makedirs()` tạo thư mục và các thư mục trung gian
- Tham số `exist_ok=True` tránh lỗi nếu thư mục đã tồn tại
- `os.listdir()` liệt kê nội dung của thư mục

## 3. Nhập và xuất tập tin

```python
filename = os.path.join(mytmpdir, "myfile.txt")
lines = ["Dans python tout est bon", "Enfin, presque"]

with open(filename, "w") as f:
    for line in lines:
        f.write(line + "\n")

with open(filename, "r") as f:
    lines = [line for line in f]
```

**Ghi chú:**
- Dùng `open(..., "w")` để ghi file
- Dùng `open(..., "r")` để đọc file
- Cú pháp `with open(...)` giúp tự động đóng file

## 4. Duyệt thư mục đệ quy

```python
for dirpath, dirnames, filenames in os.walk(mytmpdir):
    print(dirpath)
    print(dirnames)
    print(filenames)
```

**Ghi chú:**
- `os.walk()` cho phép duyệt toàn bộ cây thư mục
- Trả về: đường dẫn hiện tại, danh sách thư mục con, danh sách tập tin

## 5. Tìm kiếm tập tin theo mẫu

```python
import glob

filenames = glob.glob(os.path.join(tmpdir, "**", "*.txt"), recursive=True)
print(filenames)
```

**Ghi chú:**
- `glob.glob()` dùng để tìm file theo mẫu
- `**/*.txt` tìm tất cả file `.txt` trong các thư mục con

## Các nguyên lý lập trình hướng đối tượng

- **Encapsulation (Đóng gói):** Gom dữ liệu (attributes) và hành vi (methods) vào trong một lớp.
- **Inheritance (Kế thừa):** Lớp con kế thừa thuộc tính và phương thức của lớp cha.
- **Polymorphism (Đa hình):** Cùng một phương thức nhưng hành vi khác nhau tùy đối tượng.
- **Abstraction (Trừu tượng):** Định nghĩa giao diện chung, lớp con phải triển khai chi tiết.
  

```python
import math

# Lớp trừu tượng (Abstract base)
class Shape2D:
    def area(self):
        raise NotImplementedError()

# Inheritance + Encapsulation
class Square(Shape2D):
    def __init__(self, width):
        self.width = width

    def area(self):
        return self.width ** 2

class Disk(Shape2D):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

shapes = [Square(2), Disk(3)]

# Polymorphism
print([s.area() for s in shapes])

s = Shape2D()
try:
    s.area()
except NotImplementedError as e:
    print("NotImplementedError", e)
```

**Giải thích:**

- `Shape2D` đóng vai trò là lớp cha trừu tượng.
- Các lớp `Square`, `Disk` kế thừa từ `Shape2D`.
- Phương thức `area()` được **override** ở mỗi lớp con.
- Khi duyệt danh sách `shapes`, Python tự động gọi đúng `area()` tương ứng.


## Mở rộng: Tính chu vi (perimeter)


```python
class Shape2D:
    def area(self):
        raise NotImplementedError()

    def perimeter(self):
        raise NotImplementedError()


class Square(Shape2D):
    def __init__(self, width):
        self.width = width

    def area(self):
        return self.width ** 2

    def perimeter(self):
        return 4 * self.width


class Disk(Shape2D):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius


shapes = [Square(2), Disk(3)]

for s in shapes:
    print(f"Area = {s.area():.2f}, Perimeter = {s.perimeter():.2f}")
```

**Nhận xét:**

- Cùng một lời gọi `perimeter()` nhưng mỗi lớp có cách tính khác nhau.
- Đây là ví dụ điển hình của **Polymorphism (đa hình)**.


## Encapsulation: Thuộc tính private


```python
class Square:
    def __init__(self, width):
        self.__width = width   # private attribute

    def area(self):
        return self.__width ** 2

    def get_width(self):
        return self.__width

    def set_width(self, w):
        if w > 0:
            self.__width = w


sq = Square(4)
print("Area:", sq.area())

sq.set_width(5)
print("New area:", sq.area())
```

**Giải thích:**

- Thuộc tính `__width` là private, không truy cập trực tiếp từ bên ngoài.
- Truy cập thông qua getter và setter.
- Giúp kiểm soát dữ liệu và tránh lỗi.

## Ứng dụng: Tính tổng diện tích các hình


```python
total_area = sum(s.area() for s in shapes)
print("Tổng diện tích:", total_area)
```
